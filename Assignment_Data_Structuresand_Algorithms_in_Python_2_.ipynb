5. Write a code to find all occurrences of a given substring within another string.

```python
def find_substring_occurrences(s, sub):
    return [i for i in range(len(s) - len(sub) + 1) if s[i:i+len(sub)] == sub]
```

6. Write a code to perform basic string compression using the counts of repeated characters

```python
def string_compression(s):
    compressed = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            compressed.append(s[i-1] + str(count))
            count = 1
    compressed.append(s[-1] + str(count))
    return ''.join(compressed)
```

14. Write a code to reverse a list in-place without using any built-in reverse functions

```python
def reverse_list(lst):
    left, right = 0, len(lst) - 1
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]
        left += 1
        right -= 1
    return lst
```

15. Implement a code to find and remove duplicates from a list while preserving the original order

```python
def remove_duplicates(lst):
    seen = set()
    return [x for x in lst if x not in seen and not seen.add(x)]
```

20. Write a code to shuffle a given list randomly without using any built-in shuffle functions

```python
import random
def shuffle_list(lst):
    for i in range(len(lst)):
        j = random.randint(0, i)
        lst[i], lst[j] = lst[j], lst[i]
    return lst
```

32. Write a code to merge two dictionaries

```python
def merge_dictionaries(dict1, dict2):
    merged = dict1.copy()
    for k, v in dict2.items():
        merged[k] = merged.get(k, 0) + v
    return merged
```

33. Write a code to access a value in a nested dictionary using a list of keys.

```python
def access_nested_value(d, keys):
    try:
        for key in keys:
            d = d[key]
        return d
    except KeyError:
        return None
```

34. Write a code to sort a dictionary by its values (ascending or descending)

```python
def sort_dict(d, descending=False):
    return dict(sorted(d.items(), key=lambda x: x[1], reverse=descending))
```

35. Write a code to invert a dictionary (keys become values, values become keys).

```python
def invert_dictionary(d):
    inverted = {}
    for k, v in d.items():
        inverted.setdefault(v, []).append(k)
    return inverted
```
