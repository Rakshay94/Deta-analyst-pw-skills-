3. Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution.

```python
def bernoulli_pdf(p):
    return [1 - p, p]  # Probabilities for failure (0) and success (1)

# Example usage
p = 0.3  # Probability of success
print("Bernoulli PDF (p=0.3):", bernoulli_pdf(p))
```

Explanation:

The Bernoulli distribution has two outcomes: success (1) with probability p and failure (0) with probability 1-p.

6. Write a Python program to calculate and plot the cumulative distribution function (CDF) of a discrete uniform distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define outcomes of a fair die (1 to 6)
outcomes = np.arange(1, 7)
pmf = np.ones_like(outcomes) / len(outcomes)  # Equal probability for each outcome

# Calculate CDF
cdf = np.cumsum(pmf)

# Plot CDF
plt.step(outcomes, cdf, where='post')
plt.title("Discrete Uniform Distribution CDF")
plt.xlabel("Outcomes")
plt.ylabel("Cumulative Probability")
plt.show()
```

Explanation:

The CDF is the cumulative sum of the PMF.

A step plot is used to visualize the CDF.

9. Write a Python function to calculate Z-scores from a dataset and plot them.

```python
import numpy as np
import matplotlib.pyplot as plt

# Function to calculate Z-scores
def calculate_z_scores(data):
    return (data - np.mean(data)) / np.std(data)

# Example usage
normal_data = np.random.normal(0, 1, 1000)  # Simulate normal data
z_scores = calculate_z_scores(normal_data)

# Plot Z-scores
plt.hist(z_scores, bins=30, density=True, alpha=0.7)
plt.title("Z-Scores Distribution")
plt.xlabel("Z-Score")
plt.ylabel("Density")
plt.show()
```

Explanation:

Z-scores standardize the data by subtracting the mean and dividing by the standard deviation.

15. Implement hypothesis testing using Z-statistics for a sample dataset.

```python
# Given data
sample_mean = 72
population_mean = 70
population_std = 5
sample_size = 100

# Calculate Z-statistic
z_stat = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))
print("Z-Statistic:", z_stat)

# Compare to critical value (95% confidence level)
if abs(z_stat) > 1.96:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

Explanation:

Z-statistics are used to test hypotheses about population means.
