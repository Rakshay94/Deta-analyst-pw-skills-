1. What is hypothesis testing in statistics?

Hypothesis testing is a statistical method used to make decisions or inferences about a population based on sample data. It involves testing an assumption (hypothesis) about a population parameter (e.g., mean, proportion) to determine whether there is enough evidence to support or reject the assumption.

---



2. What is the null hypothesis, and how does it differ from the alternative hypothesis?



*   Null Hypothesis (H₀): This is the default assumption that there is no effect or no difference. It represents the status quo.

*  Alternative Hypothesis (H₁ or Ha): This is the claim we want to test. It represents the presence of an effect or a difference.

For example:

*   H₀: The mean score of a group is 50
*   H₁: The mean score of a group is not 50.

---





3. What is the significance level in hypothesis testing, and why is it important?



*   The significance level (α) is the probability of rejecting the null hypothesis when it is true (Type I error). Common values are 0.05, 0.01, or 0.10.

*  It is important because it determines the threshold for deciding whether the results are statistically significant.

---



4. What does a P-value represent in hypothesis testing?



*   The P-value is the probability of observing the sample data (or more extreme data) if the null hypothesis is true.

*   A small P-value (≤ α) suggests strong evidence against the null hypothesis, leading to its rejection.


---




5. How do you interpret the P-value in hypothesis testing?



*   If P-value ≤ α, reject the null hypothesis (evidence supports the alternative).
*   If P-value > α, fail to reject the null hypothesis (insufficient evidence to support the alternative).

---





6. What are Type 1 and Type 2 errors in hypothesis testing?



*  Type 1 Error (α): Rejecting the null hypothesis when it is true (false positive).


* Type 2 Error (β): Failing to reject the null hypothesis when it is false (false negative).

7. What is the difference between a one-tailed and a two-tailed test in hypothesis testing?



*   One-tailed test: Tests for the possibility of an effect in one direction (e.g., greater than or less than).


* Two-tailed test: Tests for the possibility of an effect in both directions (e.g., not equal to).

Question 1: Write a Python program to perform a Z-test for comparing a sample mean to a known population mean and interpret the results.

```python
import numpy as np
from scipy.stats import norm

# Given data
population_mean = 100  # Known population mean
sample_mean = 105      # Sample mean
sample_std = 15        # Sample standard deviation
sample_size = 30       # Sample size
alpha = 0.05           # Significance level

# Calculate the Z-score
z_score = (sample_mean - population_mean) / (sample_std / np.sqrt(sample_size))

# Calculate the P-value (two-tailed test)
p_value = 2 * (1 - norm.cdf(abs(z_score)))

# Interpret the results
print(f"Z-score: {z_score:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < alpha:
    print("Reject the null hypothesis: The sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: The sample mean is not significantly different from the population mean.")
```

Question 4: Perform a two-tailed Z-test using Python and visualize the decision region on a plot.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Given data
population_mean = 100
sample_mean = 105
sample_std = 15
sample_size = 30
alpha = 0.05

# Calculate Z-score
z_score = (sample_mean - population_mean) / (sample_std / np.sqrt(sample_size))

# Critical Z-values for two-tailed test
z_critical = norm.ppf(1 - alpha / 2)

# Plot the decision region
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)
plt.plot(x, y, label="Standard Normal Distribution")

# Shade rejection regions
plt.fill_between(x, y, where=(x < -z_critical) | (x > z_critical), color='red', alpha=0.5, label="Rejection Region")

# Plot Z-score
plt.axvline(z_score, color='blue', linestyle='--', label=f"Z-score = {z_score:.2f}")

# Add labels and legend
plt.title("Two-tailed Z-test Decision Region")
plt.xlabel("Z-score")
plt.ylabel("Probability Density")
plt.legend()
plt.show()
```

Question 5: Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def type1_type2_error(alpha, population_mean, sample_mean, sample_std, sample_size, true_mean):
    """
    Calculate and visualize Type 1 and Type 2 errors.
    """
    # Standard error
    se = sample_std / np.sqrt(sample_size)

    # Critical Z-value for Type 1 error
    z_critical = norm.ppf(1 - alpha / 2)

    # Type 1 error region
    x = np.linspace(population_mean - 4 * se, population_mean + 4 * se, 1000)
    y_null = norm.pdf(x, loc=population_mean, scale=se)
    plt.plot(x, y_null, label="Null Hypothesis Distribution")

    # Shade Type 1 error region
    plt.fill_between(x, y_null, where=(x < population_mean - z_critical * se) | (x > population_mean + z_critical * se),
                     color='red', alpha=0.5, label="Type 1 Error (α)")

    # Type 2 error region
    y_alt = norm.pdf(x, loc=true_mean, scale=se)
    plt.plot(x, y_alt, label="Alternative Hypothesis Distribution")

    # Shade Type 2 error region
    plt.fill_between(x, y_alt, where=(x > population_mean - z_critical * se) & (x < population_mean + z_critical * se),
                     color='blue', alpha=0.5, label="Type 2 Error (β)")

    # Add labels and legend
    plt.title("Type 1 and Type 2 Errors")
    plt.xlabel("Sample Mean")
    plt.ylabel("Probability Density")
    plt.legend()
    plt.show()

# Example usage
type1_type2_error(alpha=0.05, population_mean=100, sample_mean=105, sample_std=15, sample_size=30, true_mean=105)
```

Question 9: Write a Python function to calculate the confidence interval for a sample mean and explain its significance.

```python
import numpy as np
from scipy.stats import norm

def confidence_interval(data, confidence_level=0.95):
    """
    Calculate the confidence interval for a sample mean.
    """
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)  # ddof=1 for sample standard deviation
    sample_size = len(data)

    # Critical Z-value
    z_critical = norm.ppf((1 + confidence_level) / 2)

    # Margin of error
    margin_of_error = z_critical * (sample_std / np.sqrt(sample_size))

    # Confidence interval
    ci_lower = sample_mean - margin_of_error
    ci_upper = sample_mean + margin_of_error

    return ci_lower, ci_upper

# Example usage
data = [23, 25, 28, 30, 32]
ci_lower, ci_upper = confidence_interval(data)
print(f"95% Confidence Interval: ({ci_lower:.4f}, {ci_upper:.4f})")
```

Question 10: Write a Python program to calculate the margin of error for a given confidence level using sample data.

```python
import numpy as np
from scipy.stats import norm

def margin_of_error(data, confidence_level=0.95):
    """
    Calculate the margin of error for a given confidence level.
    """
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)  # ddof=1 for sample standard deviation
    sample_size = len(data)

    # Critical Z-value
    z_critical = norm.ppf((1 + confidence_level) / 2)

    # Margin of error
    margin_error = z_critical * (sample_std / np.sqrt(sample_size))

    return margin_error

# Example usage
data = [23, 25, 28, 30, 32]
confidence_level = 0.95
margin_error = margin_of_error(data, confidence_level)
print(f"Margin of Error (Confidence Level = {confidence_level}): {margin_error:.4f}")
```

Question 11: Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process.

```python
def bayes_theorem(prior_A, likelihood_B_given_A, likelihood_B_given_not_A):
    """
    Calculate the posterior probability using Bayes' Theorem.
    """
    # Calculate P(B)
    p_B = (likelihood_B_given_A * prior_A) + (likelihood_B_given_not_A * (1 - prior_A))

    # Calculate P(A|B) using Bayes' Theorem
    posterior_A_given_B = (likelihood_B_given_A * prior_A) / p_B

    return posterior_A_given_B

# Example usage
prior_A = 0.01  # P(A)
likelihood_B_given_A = 0.95  # P(B|A)
likelihood_B_given_not_A = 0.05  # P(B|¬A)

posterior = bayes_theorem(prior_A, likelihood_B_given_A, likelihood_B_given_not_A)
print(f"Posterior Probability P(A|B): {posterior:.4f}")
```

question 19: Write a Python function to check the assumptions (normality, independence, and equal variance) for ANOVA.

```python
import scipy.stats as stats
import numpy as np

def check_anova_assumptions(*groups):
    # Normality test using Shapiro-Wilk
    for i, group in enumerate(groups):
        stat, p_value = stats.shapiro(group)
        print(f"Group {i+1} normality test - p-value: {p_value}")

    # Equality of variance test (Levene’s Test)
    stat, p_value = stats.levene(*groups)
    print(f"Equality of variance test (Levene) p-value: {p_value}")

# Example usage
check_anova_assumptions(group1, group2, group3)
```

question 20: Perform a two-way ANOVA test using Python to study the interaction between two factors and visualize the results.

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
import seaborn as sns

# Simulate data
np.random.seed(42)
factor1 = np.random.choice(['A', 'B'], 200)
factor2 = np.random.choice(['X', 'Y'], 200)
response = np.random.normal(10, 2, 200) + np.where(factor1 == 'A', 2, 0) + np.where(factor2 == 'X', 1, 0)

# Create DataFrame
import pandas as pd
data = pd.DataFrame({'Factor1': factor1, 'Factor2': factor2, 'Response': response})

# Perform two-way ANOVA
model = ols('Response ~ Factor1 * Factor2', data=data).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

# Print results
print(anova_table)

# Plotting
sns.boxplot(x='Factor1', y='Response', hue='Factor2', data=data)
plt.title('Two-way ANOVA')
plt.show()
```
