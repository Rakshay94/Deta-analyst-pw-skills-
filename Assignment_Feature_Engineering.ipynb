**1. What is a parameter?**

A parameter is a variable that is used by a model to make predictions. Parameters are learned during the training process of a machine learning model (e.g., weights in a linear regression model)

**2. What is correlation?**

Correlation is a statistical measure that describes the strength and direction of a relationship between two variables. It ranges from -1 to 1. A correlation of 1 indicates a perfect positive relationship, -1 indicates a perfect negative relationship, and 0 indicates no relationship.
