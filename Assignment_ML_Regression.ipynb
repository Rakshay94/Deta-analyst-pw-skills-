**1. What is Simple Linear Regression?**
  

  Simple Linear Regression is a statistical method used to model the relationship between two variables:

*  One independent variable (X)
*   One dependent variable (Y)

It is represented by the equation:

           Y=mX+c

where:


*  m is the slope (rate of change of Y with respect to X)
*   c is the intercept (value of Y when X = 0)

Simple Linear Regression assumes a linear relationship between X and Y.

---





**2. What are the key assumptions of Simple Linear Regression?**



1.  Linearity: The relationship between X and Y is linear.

2.   Independence: The observations are independent of each other.

3.   Homoscedasticity: The variance of residuals is constant.
4.  Normality of Residuals: The residuals (errors) are normally distributed.


5.  No Multicollinearity: (Only applicable to multiple regression) Independent variables should not be highly correlated.

**6. What is the purpose of the least squares method in Simple Linear Regression?**

The least squares method minimizes the sum of the squared residuals (errors) to find the best-fitting line. The goal is to reduce the difference between actual and predicted values.

**7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?**



*   R² (R-squared) measures the proportion of variance in Y explained by X
*   Values range from 0 to 1

  *   0 → No explanatory power
  *  1 → Perfect fit

  Higher R² means the model explains more variation in Y.

**11. What is heteroscedasticity, and how does it affect a Multiple Linear Regression model?**



*   Heteroscedasticity means that the residuals do not have constant variance across all values of X
*   This leads to biased standard errors, affecting statistical significance.
